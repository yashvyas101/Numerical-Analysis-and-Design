## Question 1: Finding Roots of Polynomial Equations

```python
def f(x):
    return x**3 - 2 * x - 5


def df(x):
    return 3 * x**2 - 2


def g(x):
    return (2 * x + 5) ** (1 / 3)
```

### Method 1: Bisection Method

```python
def bisection_method(func, a, b, tol=.0001, max_iter=100):
    fa = func(a)
    fb = func(b)

    if fa * fb >= 0:
        raise ValueError("Bisection method requires f(a) * f(b) < 0.")

    for iteration in range(1, max_iter + 1):
        c = (a + b) / 2.0
        fc = func(c)

        if abs(fc) <= tol or (b - a) / 2.0 <= tol:
            return c, iteration

        if fa * fc < 0:
            b = c
            fb = fc
        else:
            a = c
            fa = fc

    return c, max_iter


root, iterations = bisection_method(f, 2.0, 3.0)
print("Bisection Root:", root)
print("Iterations:", iterations)
```

Time Complexity: O(log2((b-a)/tol))

Space Complexity: O(1)

### Method 2: Regula-Falsi Method

```python
def regula_falsi_method(func, a, b, tol=.0001, max_iter=100):
    fa = func(a)
    fb = func(b)

    if fa * fb >= 0:
        raise ValueError("Regula-Falsi method requires f(a) * f(b) < 0.")

    x = a
    for iteration in range(1, max_iter + 1):
        x = (a * fb - b * fa) / (fb - fa)
        fx = func(x)

        if abs(fx) <= tol:
            return x, iteration

        if fa * fx < 0:
            b = x
            fb = fx
        else:
            a = x
            fa = fx

    return x, max_iter


root, iterations = regula_falsi_method(f, 2.0, 3.0)
print("Regula-Falsi Root:", root)
print("Iterations:", iterations)
```

Time Complexity: O(k)

Space Complexity: O(1)

### Method 3: Secant Method

```python
def secant_method(func, x0, x1, tol=.0001, max_iter=100):
    f0 = func(x0)
    f1 = func(x1)

    for iteration in range(1, max_iter + 1):
        denominator = f1 - f0
        if denominator == 0:
            raise ZeroDivisionError("Secant method encountered division by zero.")

        x2 = x1 - f1 * (x1 - x0) / denominator
        f2 = func(x2)

        if abs(x2 - x1) <= tol or abs(f2) <= tol:
            return x2, iteration

        x0, f0 = x1, f1
        x1, f1 = x2, f2

    return x2, max_iter


root, iterations = secant_method(f, 2.0, 3.0)
print("Secant Root:", root)
print("Iterations:", iterations)
```

Time Complexity: O(k)

Space Complexity: O(1)

### Method 4: Newton Method

```python
def newton_method(func, dfunc, x0, tol=.0001, max_iter=100):
    x = x0

    for iteration in range(1, max_iter + 1):
        fx = func(x)
        dfx = dfunc(x)

        if dfx == 0:
            raise ZeroDivisionError("Newton method encountered zero derivative.")

        x_next = x - fx / dfx

        if abs(x_next - x) <= tol or abs(func(x_next)) <= tol:
            return x_next, iteration

        x = x_next

    return x, max_iter


root, iterations = newton_method(f, df, 2.5)
print("Newton Root:", root)
print("Iterations:", iterations)
```

Time Complexity: O(k)

Space Complexity: O(1)

### Method 5: Fixed Point Iteration Method

```python
def fixed_point_method(gfunc, x0, tol=.0001, max_iter=100):
    x = x0

    for iteration in range(1, max_iter + 1):
        x_next = gfunc(x)

        if abs(x_next - x) <= tol:
            return x_next, iteration

        x = x_next

    return x, max_iter


root, iterations = fixed_point_method(g, 2.5)
print("Fixed Point Root:", root)
print("Iterations:", iterations)
```

Time Complexity: O(k)

Space Complexity: O(1)
